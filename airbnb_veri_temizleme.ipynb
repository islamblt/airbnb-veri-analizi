# Airbnb Veri Analizi ve Veri Temizleme Projesi

## 1. Projenin Amacı

Bu projede New York City Airbnb 2019 veri seti üzerinde veri kalitesi analizi ve veri temizleme işlemleri gerçekleştirilecektir.

Çalışmanın temel amaçları:

- Veri setinin genel yapısını incelemek.
- Eksik verileri tespit etmek.
- Yinelenen kayıtları kontrol etmek.
- Hatalı ve tutarsız verileri belirlemek.
- Gerekli veri temizleme işlemlerini uygulamak.
- Temizleme öncesi ve sonrası sonuçları karşılaştırmak.

## 2. Veri Seti Hakkında

- **Veri seti:** New York City Airbnb Open Data (2019)
- **Dosya adı:** AB_NYC_2019.csv
- **Veri kaynağı:** Inside Airbnb

## 3. Kullanılan Teknolojiler

- Python
- Pandas
- NumPy
- Jupyter Notebook

```python
import pandas as pd
import numpy as np
from pathlib import Path

print("Gerekli kütüphaneler başarıyla yüklendi.")
```

```python
dosya_yolu = Path("data/AB_NYC_2019.csv")

df = pd.read_csv(dosya_yolu)

print("Veri seti başarıyla yüklendi.")
```

```python
print("Toplam kayıt (satır):", df.shape[0])
print("Toplam özellik (sütun):", df.shape[1])
```

```python
print("Veri setindeki sütunlar:\n")

for sutun in df.columns:
    print("-", sutun)
```

```python
df.info()
```

```python
eksik_veriler = df.isnull().sum()

print("Sütunlara göre eksik veri sayıları:\n")
print(eksik_veriler[eksik_veriler > 0])
```

```python
eksik_oran = (df.isnull().sum() / len(df)) * 100

eksik_raporu = pd.DataFrame({
    "Eksik Veri Sayısı": df.isnull().sum(),
    "Eksik Veri Oranı (%)": eksik_oran
})

eksik_raporu = eksik_raporu[eksik_raporu["Eksik Veri Sayısı"] > 0]

print(eksik_raporu.round(2))
```

```python
print(
    "İki sütunda da eksik olan kayıt sayısı:",
    (df["last_review"].isnull() & df["reviews_per_month"].isnull()).sum()
)
```

```python
print("Tekrarlanan satır sayısı:", df.duplicated().sum())
```

```python
df.describe().T
```

```python
print("Fiyatı 0 olan ilan sayısı:", (df["price"] == 0).sum())

print("Fiyatı 0'dan küçük olan ilan sayısı:", (df["price"] < 0).sum())

print("Fiyatı 1000 doların üzerinde olan ilan sayısı:", (df["price"] > 1000).sum())

print("\nEn yüksek 10 ilan fiyatı:")
print(df["price"].nlargest(10).to_list())
```

```python
df[df["price"] == 0][
    ["id", "name", "neighbourhood_group", "neighbourhood", "room_type", "price"]
]
```

```python
sifir_fiyat_sayisi = (df["price"] == 0).sum()

sifir_fiyat_orani = (sifir_fiyat_sayisi / len(df)) * 100

print("Sıfır fiyatlı ilan sayısı:", sifir_fiyat_sayisi)
print("Veri setindeki oranı (%):", round(sifir_fiyat_orani, 4))
```

```python
print("0 veya negatif minimum gece sayısı:",
      (df["minimum_nights"] <= 0).sum())

print("365 geceden fazla minimum gece sayısı:",
      (df["minimum_nights"] > 365).sum())

print("\nEn yüksek 10 minimum gece değeri:")
print(df["minimum_nights"].nlargest(10).to_list())
```

```python
df[df["minimum_nights"] > 365][
    ["id", "name", "neighbourhood_group", "neighbourhood",
     "room_type", "price", "minimum_nights"]
].sort_values(by="minimum_nights", ascending=False)
```

```python
print("Geçersiz enlem sayısı:",
      ((df["latitude"] < -90) | (df["latitude"] > 90)).sum())

print("Geçersiz boylam sayısı:",
      ((df["longitude"] < -180) | (df["longitude"] > 180)).sum())
```

```python
ny_disinda = df[
    (df["latitude"] < 40.4) |
    (df["latitude"] > 41.0) |
    (df["longitude"] < -74.3) |
    (df["longitude"] > -73.6)
]

print("Yaklaşık New York sınırları dışında kalan ilan sayısı:",
      len(ny_disinda))
```

```python
print("Oda türleri:")
print(df["room_type"].value_counts())

print("\nBenzersiz oda türleri:")
print(df["room_type"].unique())
```

```python
print("Bölgelere göre ilan sayıları:\n")
print(df["neighbourhood_group"].value_counts())

print("\nEksik veya boş bölge bilgisi:",
      (df["neighbourhood_group"].isnull() |
       (df["neighbourhood_group"].str.strip() == "")).sum())
```

```python
print("Negatif yorum sayısı:",
      (df["number_of_reviews"] < 0).sum())

print("Yorum sayısı 0 olan ilanlar:",
      (df["number_of_reviews"] == 0).sum())

print("En yüksek yorum sayısı:",
      df["number_of_reviews"].max())
```

```python
print(
    "Yorum sayısı 0 ve son yorum tarihi boş:",
    ((df["number_of_reviews"] == 0) &
     (df["last_review"].isna())).sum()
)

print(
    "Yorum sayısı 0 ve aylık yorum sayısı boş:",
    ((df["number_of_reviews"] == 0) &
     (df["reviews_per_month"].isna())).sum()
)
```

```python
print("0 günden az müsaitlik:",
      (df["availability_365"] < 0).sum())

print("365 günden fazla müsaitlik:",
      (df["availability_365"] > 365).sum())

print("En düşük müsaitlik:", df["availability_365"].min())
print("En yüksek müsaitlik:", df["availability_365"].max())
```

```python
print("İlan adı eksik olan kayıtlar:")
display(df[df["name"].isna()][["id", "host_id", "host_name", "neighbourhood_group"]])

print("\nEv sahibi adı eksik olan kayıtlar:")
display(df[df["host_name"].isna()][["id", "name", "host_id", "neighbourhood_group"]])
```

```python
print("Ev sahibi adı eksik olan kayıtlar:")

display(
    df[df["host_name"].isna()][
        ["id", "name", "host_id", "neighbourhood_group"]
    ]
)
```

```python
print("Negatif aylık yorum sayısı:",
      (df["reviews_per_month"] < 0).sum())

print("Sıfır aylık yorum sayısı:",
      (df["reviews_per_month"] == 0).sum())

print("En yüksek aylık yorum sayısı:",
      df["reviews_per_month"].max())
```

```python
tarihler = pd.to_datetime(df["last_review"], errors="coerce")

print("Geçerli tarih sayısı:", tarihler.notna().sum())
print("Dönüştürülemeyen tarih sayısı:",
      (df["last_review"].notna() & tarihler.isna()).sum())

print("En eski yorum tarihi:", tarihler.min())
print("En yeni yorum tarihi:", tarihler.max())
```

```python
satir_sayisi_once = len(df)
sutun_sayisi_once = len(df.columns)

print("Temizleme öncesi satır sayısı:", satir_sayisi_once)
print("Temizleme öncesi sütun sayısı:", sutun_sayisi_once)
```
